Task 1 — Unit Converter

```python
def convert_units():
    """
    Converts different units of measurement.

    Conversion Types:
    1. Length (meters ↔ feet)
    2. Weight (kilograms ↔ pounds)
    3. Volume (liters ↔ gallons)

    Returns:
        float: converted value depending on user selection
    """

    print("Select conversion type:")
    print("1. Length")
    print("2. Weight")
    print("3. Volume")

    choice = input("Enter choice (1/2/3): ")

    try:
        value = float(input("Enter value to convert: "))

        if choice == "1":
            print("1. meters → feet")
            print("2. feet → meters")
            sub = input("Choose conversion: ")

            if sub == "1":
                result = value * 3.28084
                print(f"{value} meters = {result} feet")

            elif sub == "2":
                result = value / 3.28084
                print(f"{value} feet = {result} meters")

            else:
                print("Invalid length option")

        elif choice == "2":
            print("1. kg → pounds")
            print("2. pounds → kg")
            sub = input("Choose conversion: ")

            if sub == "1":
                result = value * 2.20462
                print(f"{value} kg = {result} pounds")

            elif sub == "2":
                result = value / 2.20462
                print(f"{value} pounds = {result} kg")

            else:
                print("Invalid weight option")

        elif choice == "3":
            print("1. liters → gallons")
            print("2. gallons → liters")
            sub = input("Choose conversion: ")

            if sub == "1":
                result = value * 0.264172
                print(f"{value} liters = {result} gallons")

            elif sub == "2":
                result = value / 0.264172
                print(f"{value} gallons = {result} liters")

            else:
                print("Invalid volume option")

        else:
            print("Invalid conversion type")

    except ValueError:
        print("Invalid input! Please enter a numeric value.")


convert_units()
```

Task 2 — Mathematical Operations on List

```python
def calculate_sum(numbers):
    """
    Returns the sum of a list of numbers.

    Parameters:
    numbers (list): list of numeric values

    Returns:
    float: sum of numbers
    """
    return sum(numbers)


def calculate_average(numbers):
    """
    Returns the average of a list of numbers.

    Parameters:
    numbers (list): list of numeric values

    Returns:
    float: average value
    """
    return sum(numbers) / len(numbers)


def find_max(numbers):
    """
    Returns the maximum value in a list.

    Parameters:
    numbers (list): list of numeric values

    Returns:
    float: maximum value
    """
    return max(numbers)


def find_min(numbers):
    """
    Returns the minimum value in a list.

    Parameters:
    numbers (list): list of numeric values

    Returns:
    float: minimum value
    """
    return min(numbers)


def main():
    print("Choose an operation:")
    print("1. Sum")
    print("2. Average")
    print("3. Maximum")
    print("4. Minimum")

    choice = input("Enter your choice (1-4): ")

    try:
        numbers = input("Enter numbers separated by space: ")
        numbers_list = [float(num) for num in numbers.split()]

        if len(numbers_list) == 0:
            print("List cannot be empty")
            return

        if choice == "1":
            print("Sum =", calculate_sum(numbers_list))

        elif choice == "2":
            print("Average =", calculate_average(numbers_list))

        elif choice == "3":
            print("Maximum =", find_max(numbers_list))

        elif choice == "4":
            print("Minimum =", find_min(numbers_list))

        else:
            print("Invalid operation selected")

    except ValueError:
        print("Invalid input! Please enter numeric values only.")


main()
```

4.3 Exercise on Nested List
1 Flatten Nested List

```python
def flatten(lst):
    result = []
    for sub in lst:
        for item in sub:
            result.append(item)
    return result

print(flatten([[1,2],[3,4],[5]]))
```

2 Access Nested Element

```python
def access_nested_element(lst, indices):
    element = lst
    for i in indices:
        element = element[i]
    return element

lst = [[1,2,3],[4,5,6],[7,8,9]]
print(access_nested_element(lst,[1,2]))
```

3 Sum of Nested List

```python
def sum_nested(lst):
    total = 0
    for item in lst:
        if isinstance(item,list):
            total += sum_nested(item)
        else:
            total += item
    return total

print(sum_nested([[1,2],[3,[4,5]],6]))
```

4 Remove Element from Nested List

```python
def remove_element(lst, elem):
    result = []
    for item in lst:
        if isinstance(item,list):
            result.append(remove_element(item,elem))
        elif item != elem:
            result.append(item)
    return result

print(remove_element([[1,2],[3,2],[4,5]],2))
```

5 Find Maximum Element

```python
def find_max(lst):
    maximum = float('-inf')

    for item in lst:
        if isinstance(item,list):
            maximum = max(maximum, find_max(item))
        else:
            maximum = max(maximum,item)

    return maximum

print(find_max([[1,2],[3,[4,5]],6]))
```

6 Count Occurrences

```python
def count_occurrences(lst, elem):
    count = 0
    for item in lst:
        if isinstance(item,list):
            count += count_occurrences(item,elem)
        elif item == elem:
            count += 1
    return count

print(count_occurrences([[1,2],[2,3],[2,4]],2))
```

7 Deep Flatten

```python
def deep_flatten(lst):
    result = []
    for item in lst:
        if isinstance(item,list):
            result.extend(deep_flatten(item))
        else:
            result.append(item)
    return result

print(deep_flatten([[[1,2],[3,4]],[[5,6],[7,8]]]))
```

8 Nested List Average

```python
def average_nested(lst):
    flat = deep_flatten(lst)
    return sum(flat)/len(flat)

print(average_nested([[1,2],[3,4],[5,6]]))
```

```python
# 1 Array from 10 to 49
arr1 = np.arange(10,50)
print(arr1)

# 2 3x3 matrix 0-8
arr2 = np.arange(9).reshape(3,3)
print(arr2)

# 3 Identity matrix
arr3 = np.eye(3)
print(arr3)

# 4 Random array size 30 and mean
arr4 = np.random.random(30)
print("Mean:", arr4.mean())

# 5 Random 10x10 matrix min and max
arr5 = np.random.random((10,10))
print("Min:", arr5.min())
print("Max:", arr5.max())

# 6 Zero array size 10 replace 5th element
arr6 = np.zeros(10)
arr6[4] = 1
print(arr6)

# 7 Reverse array
arr7 = np.array([1,2,0,0,4,0])
print(arr7[::-1])

# 8 2D array border 1 inside 0
arr8 = np.ones((5,5))
arr8[1:-1,1:-1] = 0
print(arr8)

# 9 Checkerboard 8x8
arr9 = np.zeros((8,8),dtype=int)
arr9[1::2,::2] = 1
arr9[::2,1::2] = 1
print(arr9)
```

```python
x = np.array([[1,2],[3,5]])
y = np.array([[5,6],[7,8]])

v = np.array([9,10])
w = np.array([11,12])

# 1 Add arrays
print(x + y)

# 2 Subtract arrays
print(x - y)

# 3 Multiply with integer
print(x * 2)

# 4 Square each element
print(np.square(x))

# 5 Dot products
print("v dot w =", np.dot(v,w))
print("x dot v =", np.dot(x,v))
print("x dot y =\n", np.dot(x,y))

# 6 Concatenate
print(np.concatenate((x,y),axis=0))  # row
print(np.vstack((v,w)))  # column

# 7 Concatenate x and v
print(np.vstack((x, v)))
```

Problem 4 — Matrix Operations

```python
A = np.array([[3,4],[7,8]])
B = np.array([[5,3],[2,1]])

# 1 Prove A * A^-1 = I
A_inv = np.linalg.inv(A)
print("A * A^-1 =\n", np.dot(A,A_inv))

# 2 Prove AB ≠ BA
print("AB =\n", np.dot(A,B))
print("BA =\n", np.dot(B,A))

# 3 Prove (AB)^T = B^T A^T
left = np.transpose(np.dot(A,B))
right = np.dot(np.transpose(B), np.transpose(A))

print("Left:\n", left)
print("Right:\n", right)
```

Solve Linear Equations

```python
A = np.array([
    [2,-3,1],
    [1,-1,2],
    [3,1,-1]
])

B = np.array([-1,-3,9])

# Solve AX = B
A_inv = np.linalg.inv(A)
X = np.dot(A_inv,B)

print("Solution:", X)
```
